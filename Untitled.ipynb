```python
#Downloading and preparing the Amharic dataset

from datasets import load_dataset

ds = load_dataset("wikimedia/wikipedia", "20231101.am", split="train")
with open("amharic.txt", "w", encoding="utf-8") as f:
    for row in ds:
        f.write(row["text"] + "\n")

import os
print(os.getcwd(), os.path.getsize("amharic.txt")/1e6, "MB")
```

```python
#Amharic text preprocessing

import re, torch
text = open("amharic.txt", encoding="utf-8").read()
text = re.sub(r"[A-Za-z0-9]+", "", text)
text = re.sub(r"\s+", " ", text).strip()
print(len(text), text[:200])
```

```python
#Character-level tokenization and data preparation

chars = sorted(set(text))
stoi = {c:i for i,c in enumerate(chars)}
itos = {i:c for c,i in stoi.items()}
encode = lambda s: [stoi[c] for c in s if c in stoi]
decode = lambda l: "".join(itos[i] for i in l)

data = torch.tensor(encode(text))
n = int(0.9*len(data))
train_data, val_data = data[:n], data[n:]   # 90% train / 10% validation
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
block_size, batch_size = 128, 64

def get_batch(split):
    d = train_data if split=="train" else val_data
    ix = torch.randint(len(d)-block_size-1, (batch_size,))
    x = torch.stack([d[i:i+block_size] for i in ix])
    y = torch.stack([d[i+1:i+block_size+1] for i in ix])
    return x.to(device), y.to(device)
```

```python
import torch.nn as nn, torch.nn.functional as F

class CausalSelfAttention(nn.Module):
    def __init__(s, d, heads):
        super().__init__()
        s.h = heads
        s.qkv = nn.Linear(d, 3*d)      # makes Query, Key, Value
        s.proj = nn.Linear(d, d)
        s.register_buffer("mask", torch.tril(torch.ones(block_size, block_size)))
    def forward(s, x):
        B,T,C = x.shape
        q,k,v = s.qkv(x).split(C, dim=2)
        q,k,v = [t.view(B,T,s.h,C//s.h).transpose(1,2) for t in (q,k,v)]
        att = (q @ k.transpose(-2,-1)) / (k.size(-1)**0.5)       # similarity scores
        att = att.masked_fill(s.mask[:T,:T]==0, float("-inf"))   # CAUSAL: can't see future
        att = F.softmax(att, dim=-1)
        out = (att @ v).transpose(1,2).reshape(B,T,C)            # merge heads
        return s.proj(out)

class Block(nn.Module):
    def __init__(s, d, heads):
        super().__init__()
        s.ln1, s.ln2 = nn.LayerNorm(d), nn.LayerNorm(d)
        s.attn = CausalSelfAttention(d, heads)
        s.ffn = nn.Sequential(nn.Linear(d,4*d), nn.GELU(), nn.Linear(4*d,d))
    def forward(s, x):
        x = x + s.attn(s.ln1(x))   # residual connection
        x = x + s.ffn(s.ln2(x))    # residual connection
        return x

class GPT(nn.Module):
    def __init__(s, vocab, d=256, heads=4, layers=4):
        super().__init__()
        s.tok = nn.Embedding(vocab, d)          # token embeddings
        s.pos = nn.Embedding(block_size, d)     # positional info
        s.blocks = nn.Sequential(*[Block(d,heads) for _ in range(layers)])
        s.ln = nn.LayerNorm(d)
        s.head = nn.Linear(d, vocab)            # scores for every possible next token
    def forward(s, idx, targets=None):
        B,T = idx.shape
        x = s.tok(idx) + s.pos(torch.arange(T, device=idx.device))
        logits = s.head(s.ln(s.blocks(x)))
        loss = None if targets is None else F.cross_entropy(logits.view(-1, logits.size(-1)), targets.view(-1))
        return logits, loss
```

```python
model = GPT(len(chars)).to(device)
opt = torch.optim.AdamW(model.parameters(), lr=3e-4)

@torch.no_grad()
def val_loss():
    model.eval()
    l = sum(model(*get_batch("val"))[1].item() for _ in range(20))/20
    model.train(); return l

for step in range(3000):
    x,y = get_batch("train")
    _, loss = model(x,y)
    opt.zero_grad(); loss.backward(); opt.step()
    if step % 250 == 0:
        print(step, "train", loss.item(), "val", val_loss())

torch.save({"model": model.state_dict(), "chars": chars}, "amharic_gpt.pt")   # save
# Load later: model.load_state_dict(torch.load("amharic_gpt.pt")["model"])
```

```python
@torch.no_grad()
def generate(prompt, n=300, temp=0.8):
    model.eval()
    idx = torch.tensor([encode(prompt)], device=device)
    for _ in range(n):
        logits,_ = model(idx[:, -block_size:])
        probs = F.softmax(logits[:, -1, :] / temp, dim=-1)
        idx = torch.cat([idx, torch.multinomial(probs, 1)], dim=1)
    return decode(idx[0].tolist())

print(generate("ኢትዮጵያ"))
```
